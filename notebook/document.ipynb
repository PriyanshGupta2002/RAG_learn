```python
sample_texts={
    "../data/text_files/python_intro.txt":"""Python Programming Introduction

Python is a high-level, interpreted programming language known for its simplicity and readability.
Created by Guido van Rossum and first released in 1991, Python has become one of the most popular
programming languages in the world.

Key Features:
- Easy to learn and use
- Extensive standard library
- Cross-platform compatibility
- Strong community support

Python is widely used in web development, data science, artificial intelligence, and automation.""",
    
    "../data/text_files/machine_learning.txt": """Machine Learning Basics

Machine learning is a subset of artificial intelligence that enables systems to learn and improve
from experience without being explicitly programmed. It focuses on developing computer programs
that can access data and use it to learn for themselves.

Types of Machine Learning:
1. Supervised Learning: Learning with labeled data
2. Unsupervised Learning: Finding patterns in unlabeled data
3. Reinforcement Learning: Learning through rewards and penalties

Applications include image recognition, speech processing, and recommendation systems
    
    
    """

}

for filepath,content in sample_texts.items():
    with open(filepath,'w',encoding="utf-8") as f:
        f.write(content)

print("✅ Sample text files created!")
```

```python
### TextLoader
from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/text_files/python_intro.txt",encoding="utf-8")
document = loader.load()
print(document)


```

```python
### DirectoryLoader
from langchain_community.document_loaders import DirectoryLoader
dir_loader = DirectoryLoader("../data/text_files", glob="**/*.txt", loader_kwargs={"encoding": "utf-8"}, show_progress=False)
documents = dir_loader.load()
documents
```

```python
### Pdf Loader
from langchain_community.document_loaders import PyPDFLoader,PyMuPDFLoader
dir_loader_pdf = DirectoryLoader("../data/pdf_files", glob="**/*.pdf",loader_cls=PyMuPDFLoader,show_progress=False)
pdf_documents = dir_loader_pdf.load()
type(pdf_documents[0])
```

```python
### Excel Loader
from langchain_community.document_loaders import UnstructuredExcelLoader
loader = UnstructuredExcelLoader(file_path="../data/excel_files/PQE+UPI+Data.xlsx", mode="elements")

# Load the documents
documents = loader.load()
documents

for i, doc in enumerate(documents):
    print(f"--- Document {i+1} ---")
    print(f"Page Content: {doc.page_content}")
    print(f"Metadata: {doc.metadata}")
    print("-" * 30)
# dir_loader_excel = DirectoryLoader("../data/excel_files", glob="**/*.xlsx", loader_cls=UnstructuredExcelLoader, show_progress=False)
# excel_documents = dir_loader_excel.load()
# excel_documents
```

```python
### Excel Loader
from langchain_community.document_loaders import UnstructuredExcelLoader

dir_loader_excel = DirectoryLoader("../data/excel_files", glob="**/*.xlsx", loader_cls=UnstructuredExcelLoader, show_progress=False)
excel_documents = dir_loader_excel.load()
excel_documents
```

```python
### HTML Loader
from langchain_community.document_loaders import UnstructuredHTMLLoader

dir_loader_html = DirectoryLoader("../data/html_files", glob="**/*.html", loader_cls=UnstructuredHTMLLoader, show_progress=False)
html_documents = dir_loader_html.load()
html_documents

for i, doc in enumerate(html_documents):
    print(f"--- Document {i+1} ---")
    print(f"Page Content: {doc.page_content}")
    print(f"Metadata: {doc.metadata}")
```
